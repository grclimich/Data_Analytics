✔ obtenir_dades(campo)

1. Recibe un parámetro campo que indica qué dato se va a solicitar.

2. Consulta un diccionario interno con los mensajes de texto correspondientes a cada opción (como nombre, teléfono, experiencia, catalán, excel y titulación).

3. Utiliza la función input() para mostrar el mensaje por pantalla, captura la respuesta del usuario y le aplica .strip() para eliminar los espacios en blanco sobrantes.

4. Devuelve la cadena de texto limpia resultante.

```python
def obtenir_dades(campo):
    mensajes = {
        "nombre": "Ingresa tu nombre y apellidos: ",
        "telefono": "Teléfono de contacto (9 dígitos): ",
        "experiencia": "Años de experiencia?: ",
        "catalan": "Nivel de català? elige una opcion (0-3):\n 0 = sense coneixement \n 1 = fins a nivell B1 \n 2 = nivell B2 \n 3 = C1 o superior\n> ",
        "excel": "Cual es tu dominio en Excel? elige una opcion (0-3):\n 0 = nivell bàsic \n 1 = ús de funcions \n 2 = taules dinàmiques \n 3 = Visual Basic i Macros\n> ",
        "titulacion": "Que titulación tienes? elige una opcion (0-3):\n 0 = ESO \n 1 = FP \n 2 = Universitat \n 3 = Màster o superior\n> "
    }
    return input(mensajes[campo]).strip()

#pasar todo lo que no sea input a el main()
```

✔ `validar_dades(campo)`

1. Usamos un parametro global "campo" para igualar a cada campo de la función "obtenir_dades()"

2. Usamos while True para cada validación de manera que no avanza al siguiente campo hasta que no sea verdadero.

3. Las validaciones tratan de:
    - Verificar si el nombre esta vacio y no permita simbolos o numeros.
    - Que el telefono no tenga menos de 9 digitos
    - Que la experiencia tenga que ser un numero y valide un rango  entre 0 y 50 años
    - Que en la selección de habilidades se elija un numero de nivel entre 0 y 3.
   
4. Si falla algún punto anterior entonces mostrara diferentes mensajes de error segun el caso.

```python
import re #modulo de expresiones regulares para validar los simbolos en el campo nombre

def validar_dades(campo, texto):
    """Valida el texto recibido y devuelve una tupla (es_valido, valor_convertido_o_error)."""
    texto = str(texto).strip()

    if campo == "nombre":
        if texto == "":
            return False, "el nombre no puede estar vacío."
        if not re.fullmatch(r"[a-zA-ZáéíóúÁÉÍÓÚàèìòùÀÈÌÒÙäëïöüÄËÏÖÜñÑçÇ\s-]+", texto):
            return False, "el nombre no puede contener números ni símbolos."
        return True, texto

    elif campo == "telefono":
        if texto == "":
            return False, "el teléfono no puede estar vacío."
        if " " in texto:
            return False, "el teléfono no puede contener espacios en blanco."
        if len(texto) != 9:
            return False, "el formato del teléfono es incorrecto (debe tener 9 dígitos)."
        try:
            return True, int(texto)
        except ValueError:
            return False, "el teléfono debe contener solo números."

    elif campo == "experiencia":
        try:
            exp = int(texto)
            if 0 <= exp <= 50:
                return True, exp
            return False, "la experiencia debe estar entre 0 y 50 años."
        except ValueError:
            return False, "debes escribir un número entero."

    elif campo in ("catalan", "excel", "titulacion"):
        try:
            opcion = int(texto)
            if opcion in (0, 1, 2, 3):
                return True, opcion
            return False, "debes escribir una opción entre 0 y 3."
        except ValueError:
            return False, "debes escribir un número entero."
```

✔ `calcular_puntuacio(punt_cat, punt_exc,pun_tit,exp)`

1. Calcula la suma de puntaje del candidato (puntaje catalan, puntaje excel, puntaje titulación)
2. Se crea Anys = 1.5 es la variable que multiplicamos por la experiencia ingresada por el candidato.
3. Retorna la multiplicacion de (experiencia por anys) + la suma de las puntuaciones.

```python
def calcular_puntuacio(punt_cat,punt_exc,pun_tit,exp):
  suma=punt_cat+punt_exc+pun_tit
  anys=1.5
  puntuacion_total=(exp*anys)+suma
  return puntuacion_total
```

✔ `classificar_candidat(puntuacio)`

1. Recibe como parámetro la puntuación numérica del candidato (puntuacio).
2. Evalúa el valor mediante una estructura condicional (if-elif-else):
3. Si la puntuación es mayor o igual a 12, asigna el perfil "Prioritat alta".
4. Si está entre 8 y 11 (inclusive), asigna "Candidat adequat".
5. Si es menor que 8, asigna "No prioritari".

```python
def classificar_candidat(puntuacio):
  if puntuacio >=12:
    perfil_candidato="Prioritat alta"
  elif 8<= puntuacio <12:
    perfil_candidato="Candidat adequat"
  else:
    perfil_candidato="No prioritari"

  return perfil_candidato
```

✔ `mostrar_informe(lista_candidatos)`

1. Verificamos si la lista está vacía, si es así, muestra un aviso por pantalla y finaliza la función.

2. Crea una lista temporal (lista_ordenada) asignando un peso numérico a cada candidato según su clasificación (3 para prioridad alta, 2 para adecuado y 1 para no prioritario).

3. Ordena los candidatos de mayor a menor prioridad combinando el peso de la clasificación y la puntuación numérica.

4. Muestra un encabezado formateado y recorre la lista ordenada para imprimir los datos principales de cada candidato (nombre, teléfono, puntuación y clasificación) numerados del 1 en adelante.

```python
def mostrar_informe(lista_candidatos):
    """Muestra un resumen global en forma de tabla de todos los registrados."""
    if not lista_candidatos:
        print("\n[!] No hay candidatos registrados en el sistema.")
        return

     #Creamos una lista temporal que incluya el peso para poder ordenarla
    lista_ordenada = []
    for candidato in lista_candidatos:
        clasificacion = candidato[10]
        if clasificacion == "Prioritat alta":
            peso = 3
        elif clasificacion == "Candidat adequat":
            peso = 2
        else:
            peso = 1

        puntuacion = float(candidato[9])
        lista_ordenada.append((peso, puntuacion, candidato))
        # Ordenamos la lista por peso luego luego por puntuación de mayor a menor
        lista_ordenada.sort(reverse=True)

    print("\n" + "="*55)
    print(f"LISTA ORDENADA DE CANDIDATOS POR PRIORIDAD ({len(lista_candidatos)} registrados)")
    print("="*55)

    # Recorremos la lista e imprimimos del candidato [posición 2]
    for indice, item in enumerate(lista_ordenada, start=1):
        candidato = item[2]
        print(f"\n--- CANDIDATO Nº {indice} ---")
        print(f"NOMBRE COMPLETO: {candidato[0]}\nTELEFONO: {candidato[1]}\nPUNTUACIÓN: {candidato[9]} pts\nCLASIFICACIÓN: {candidato[10]}\n") # imprimimos nombre completo, telefono y puntuación
        print("-" * 40)
    print("="*40 + "\n")
```

```python
import pandas as pd

clients = [
    {"id_client": 1, "nom": "Laura Serra", "edat": 23, "grup_edat": "18-25", "codi_comportament": 4},
    {"id_client": 2, "nom": "Marc Vidal", "edat": 35, "grup_edat": "26-40", "codi_comportament": 3},
    {"id_client": 3, "nom": "Ana López", "edat": 29, "grup_edat": "26-40", "codi_comportament": 2},
    {"id_client": 4, "nom": "Joan Riera", "edat": 19, "grup_edat": "18-25", "codi_comportament": 2},
    {"id_client": 5, "nom": "Mónica Pérez", "edat": 47, "grup_edat": "41-60", "codi_comportament": 4},
    {"id_client": 6, "nom": "Luis García", "edat": 52, "grup_edat": "41-60", "codi_comportament": 1},
    {"id_client": 7, "nom": "Pilar Sánchez", "edat": 61, "grup_edat": "60+", "codi_comportament": 3},
    {"id_client": 8, "nom": "Eva Martín", "edat": 38, "grup_edat": "26-40", "codi_comportament": 0},
    {"id_client": 9, "nom": "Diego Romero", "edat": 24, "grup_edat": "18-25", "codi_comportament": 1},
    {"id_client": 10, "nom": "Núria Costa", "edat": 33, "grup_edat": "26-40", "codi_comportament": 4}
]

diccionario_comportament={
0: {"segment":"Navega pero no añade productos", "factor_codi": 0.10,"mensaje":"Descubre las ultimas novedades exclusivas de la semana."},
1: {"segment":"Agrega productos pero no llega al checkout","factor_codi": 0.25,"mensaje":"Tus artículos en cesta te esperan."},
2: {"segment":"Llega al checkout pero no paga","factor_codi": 0.50,"mensaje":"Completa tu pedido ahora y disfruta de envío gratuito."},
3: {"segment":"Cesta de compra abandonado (<24h)","factor_codi": 0.75,"mensaje":"¡Date prisa! Tu cesta de la compra caducará muy pronto."},
4: {"segment":"Cesta de compra abandonado (>24h)","factor_codi": 0.90,"mensaje":"Te ofrecemos un cupon del 10% de descuento en tu cesta de compra"}
}

diccionario_edad = {
"18-25": 1.20,
"26-40": 1.00,
"41-60": 0.85,
"60+": 0.70
}
```

Función: obtenir_segment_i_factor(codigo_comportament)
1. Recibe como parámetro el código de comportamiento numérico del cliente y realiza una consulta directa al diccionario maestro diccionario_comportament.

2. Retorna el sub-diccionario correspondiente que incluye de manera estructurada el segmento comercial, el factor de código numérico y el mensaje publicitario personalizado para ese comportamiento específico.

```python
def obtenir_segment_i_factor(codigo_comportament):
    """Retorna el sub-diccionario de código de comportamiento."""
    return diccionario_comportament[codigo_comportament]
```

Función: obtenir_factor_edat(grupo_edad)
1. Recibe el texto del grupo de edad del cliente (por ejemplo, "18-25") y consulta el diccionario de factores demográficos diccionario_edad.

2. Retorna el valor numérico del factor multiplicador asociado a dicha franja de edad para ponderar el cálculo posterior del riesgo.

```python
def obtenir_factor_edat(grupo_edad):
    """Retorna el grupo de edad."""
    return diccionario_edad[grupo_edad]
```

Función: calcular_risc(factor_codigo, factor_edat)
1. Recibe el factor de código de comportamiento y el factor de edad previamente obtenidos.

2. Multiplica ambos valores y aplica una función de redondeo a 3 decimales para calcular y retornar el riesgo final (risc_final) del cliente.

```python
def calcular_risc(factor_codigo, factor_edat):
    """Retorna la multiplicacion del codigo del factor con el factor edad"""
    return round(factor_codigo*factor_edat,3)
```

Función: processar_clientes(lista_clients)
1. Recorre iterativamente la lista completa de clientes iniciales para procesar sus datos de forma individual.

2. Interactúa directamente con las funciones auxiliares llamando a obtenir_segment_i_factor() y obtenir_factor_edat(), y posteriormente calcula el valor del riesgo utilizando la función calcular_risc().

3. Empaqueta toda la información consolidada en un diccionario detallado por cliente (nombre, ID, edad, segmento, mensaje, factores y riesgo final) y los acumula en una lista.

4. Convierte la lista resultante en un DataFrame de pandas, lo ordena de mayor a menor según el risc_final y lo retorna listo para su análisis.

```python
def processar_clientes(lista_clients):
    clients_procesados=[]

    for cliente in lista_clients:
        clients_procesados.append({
        "id_client": cliente["id_client"],
        "Nombre": cliente["nom"],
        "edat": cliente["edat"],
        "grup_edat": cliente["grup_edat"],
        "codi_comportament": cliente["codi_comportament"],
        "segment": obtenir_segment_i_factor(cliente["codi_comportament"])["segment"],
        "missatge": obtenir_segment_i_factor(cliente["codi_comportament"])["mensaje"],
        "factor_código": obtenir_segment_i_factor(cliente["codi_comportament"])["factor_codi"],
        "factor_edat": obtenir_factor_edat(cliente["grup_edat"]),
        "risc_final": calcular_risc(obtenir_segment_i_factor(cliente["codi_comportament"])["factor_codi"], obtenir_factor_edat(cliente["grup_edat"]))
        })

    df= pd.DataFrame(clients_procesados)
    return df.sort_values(by="risc_final",ascending=False)
```

## Exercici 1: Gestió de risc i priorització de pacients


✔ Estructura de datos:

- pacients: Una lista de cadenas de texto donde cada elemento representa a un paciente con el formato "Nombre|Edad|Síntomas".

- nivells_risc: Un diccionario que asigna una puntuación numérica de gravedad a cada síntoma específico.

```python
import pandas as pd

pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]

nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

✔ `limpiar_y_normalizar_sintomas`

1. Divide el texto por cada coma (,).
2. Convierte todo a minúsculas para evitar problemas de mayúsculas.
3. Elimina los espacios en blanco sobrantes a los lados (.strip()).
4. Reemplaza los espacios internos por guiones bajos (_) para que coincidan exactamente con las claves del diccionario de riesgos (por ejemplo, "fiebre alta" se convierte en "fiebre_alta").

```python
def limpiar_sintomas(sintomas): #cambiar nombres de funcion mas acorde
    """ Esta función se encarga de recibir la cadena de síntomas en bruto de un paciente (separados por comas) y
     Limpia y normaliza la cadena de síntomas transformando a minúsculas, elimina espacios y formato clave."""
    sintomas_sin_limpiar = sintomas.split(",")
    lista_sintomas = []
    for limpieza in sintomas_sin_limpiar:
        # Pasa a minúsculas, quita espacios extremos y reemplaza espacios internos por guiones bajos
        sintoma_limpio = limpieza.strip().lower().replace(" ", "_")
        lista_sintomas.append(sintoma_limpio)
    return lista_sintomas
```

✔ `calcular_riesgo()`
- Recorre la lista de síntomas ya normalizados de un paciente individual y consulta el diccionario diccionario_riesgos para sumar la puntuación correspondiente de cada síntoma, devolviendo la carga de riesgo total acumulada.

```python
def calcular_riesgo(sintomas_transformados,diccionario_riesgos):
    """obtiene el riesgo total contando la cantidad de sintomas por paciente."""
    riesgo_total=0
    for sintoma in sintomas_transformados:
        riesgo_sintoma=diccionario_riesgos.get(sintoma,0)
        riesgo_total+=riesgo_sintoma
    return riesgo_total
```

✔ `procesar_pacientes (lista_pacientes, diccionario_riesgos)`

1. Se crea una lista vacia e itera sobre cada registro de la lista de pacientes con un for.
2. Separa los campos usando el carácter delimitador vertical (|).
3. Convierte la edad a formato numérico entero (int).
4. Llama a limpiar_y_normalizar_sintomas y a calcular_riesgo.
5. Almacena toda la información estructurada en un diccionario por paciente dentro de una lista general.
6. Convierte la estructura final en un DataFrame de pandas y lo ordena de mayor a menor según el nivel de riesgo calculado.

```python
def procesar_pacientes(lista_pacientes,diccionario_riesgos):
    """funcion principal que procesa y ordena la lista de pacientes llama a una funcion para separar los sintomas de manera ordenada,
    usa el caracter especial para separar el texto (|) y procesa los datos del diccionario de riesgos."""
    pacientes_procesados=[]

    for separador in lista_pacientes:
        partes=separador.split("|")
        nombre=partes[0].strip()
        edad=int(partes[1].strip())
        sintomas_sin_formato=partes[2].strip()

        sintomas_transformados=limpiar_sintomas(sintomas_sin_formato)
        riesgo_total=calcular_riesgo(sintomas_transformados,diccionario_riesgos)

        pacientes_procesados.append({
        "Nombre": nombre,
        "Edad": edad,
        "Sintomas_normalizados": sintomas_transformados,
        "Riesgo_total": riesgo_total
    })

    df=pd.DataFrame(pacientes_procesados)
    return df.sort_values(by="Riesgo_total", ascending=False)
```

✔ `main_pacientes_normalizados()`
- Ejecuta el script pasándole los datos iniciales para obtener la tabla final con los pacientes priorizados por urgencia o nivel de riesgo.

```python
def main_pacientes_normalizados(pacientes,riesgos):
    resultado_final=procesar_pacientes(pacientes,riesgos)
    return resultado_final
```

## Exercici 2: Classificació de clients de segurs

Estructura de datos:

- clients: Una lista de diccionarios que contiene los datos personales básicos de cada cliente (nom, edat, codi de riesgo).

- nivells_seguro: Un diccionario jerárquico que relaciona los códigos de riesgo con su respectiva categoría descriptiva, precio base y probabilidad de fraude estimada.

```python

clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]


nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}
```

`calculo_precio_final()`
Calcula el precio final del seguro aplicando un incremento porcentual sobre la tarifa base en función directa de la edad del cliente.

```python
def calculo_precio_final(p_base,edad):
    """Calcula el precio final incrementando la base en función de la edad.
    Argumentos:
    p_base:
    edad:
    devuelve precio final:
    """
    preu_final = p_base * (1 + edad/100)
    return preu_final
```

`obtener_nivells_seguro()`
- Actúa como una capa de consulta estructurada para extraer la información de riesgo correspondiente a un código específico utilizando el diccionario global.

```python
def obtener_nivells_seguro(codi):
    """Devuelve el sub-diccionario correspondiente al código de riesgo.
    Argumentos:
    parametro de codigo de clientte: codi
    """
    return nivells_seguro[codi]
```

`clasificar_clientes()`
- Recorre la lista de clientes con un for, extrae los parámetros de riesgo mediante funciones auxiliares, calcula los precios dinámicos y genera un DataFrame de pandas ordenado de forma descendente por Probabilidad de fraude y Precio final.

```python
def clasificar_clientes(lista_clientes):
    """Procesa la lista de clientes, calcula los precios y devuelve un DataFrame ordenado."""
    clientes_procesados=[]

    for cliente in lista_clientes:
        info_cat = obtener_nivells_seguro(cliente["codi"])

        categoria=info_cat["categoria"]
        precio_base=int(info_cat["preu_base"])
        tipo_fraude=float(info_cat["fraude"])
        precio_final = calculo_precio_final(info_cat["preu_base"], cliente["edat"])

        clientes_procesados.append({
            "Nombre": cliente["nom"],
            "Edad":cliente["edat"],
            "Categoria": categoria,
            "Precio_Base": precio_base,
            "Precio_final": precio_final,
            "Probabilidad_fraude": tipo_fraude
        })

    df=pd.DataFrame(clientes_procesados)
    return df.sort_values(by=["Probabilidad_fraude", "Precio_final"], ascending=[False, False])
```

```python
def main_clientes_riesgo(clientes):
    """Función principal que ejecuta la clasificación y análisis."""
    result=clasificar_clientes(clientes)
    return result
```

## Exercici 1: Generador de contrasenyes segures


`crear_contrasenya(longitud, mayusculas, minusculas, numeros, signos)`

1. crea la lista inicial de conjuntos activos (lista_contrasenya).

2. Si mayusculas, minusculas, numeros o signos son verdaderos, añade sus respectivos catálogos de string a la lista de conjuntos.

3. Se inicializa una lista vacía llamada contrasenya para ir acumulando los caracteres obligatorios:

4. Recorre cada grupo activo con un bucle for y extrae un carácter aleatorio obligatorio mediante np.random.choice().

5. Se calcula el tamaño restante (resto_caracteres) restando la longitud deseada menos los caracteres obligatorios ya añadidos.

6. Se completa la contraseña seleccionando el resto de caracteres al azar del pool total combinado (caracteres) utilizando np.random.choice().

7. np.random.permutation(contrasenya) desordena por completo los elementos de la lista para evitar que los caracteres obligatorios queden siempre en las primeras posiciones.

8. Devuelve el resultado convertido en una cadena de texto mediante "".join().

```python
import numpy as np
import string

def crear_contrasenya(longitud=12,mayusculas=True,minusculas=True,numeros=True,signos=False):
    lista_contrasenya=[]

    if mayusculas:
        lista_contrasenya.append(string.ascii_uppercase)

    if minusculas:
        lista_contrasenya.append(string.ascii_lowercase)

    if numeros:
        lista_contrasenya.append(string.digits)

    if signos:
        lista_contrasenya.append(string.punctuation)

    #ahora creamos una lista para añadir los caracteres de la lista y añadimos uno a uno con un for
    contrasenya=[]

    for grupo in lista_contrasenya:
          contrasenya.append(np.random.choice(list(grupo)))

    caracteres = "".join(lista_contrasenya)
    resto_caracteres= longitud-len(contrasenya)

    contrasenya += list(np.random.choice(list(caracteres),resto_caracteres))
    contrasenya = "".join(np.random.permutation(contrasenya))

    return contrasenya
```

`cargar_partidos(archivo_texto)`

- Lee el archivo de texto separa los datos por tabuladores y construye un DataFrame inicial con las columnas local, resultado y visitante.

```python

def cargar_partidos(archivo_texto):
    """ Carga los datos del archivo de texto al DataFrame."""
    partidos=[]

    with open(archivo_texto, "r", encoding="utf-8") as fichero:

        for linea in fichero:

            datos=linea.strip().split("\t")
            partidos.append(datos)

    df=pd.DataFrame(partidos,columns=["local","resultado","visitante"])
    return df
```

`separar_goles(df)`

- Divide la columna resultado (ej. "2-1") en dos columnas numéricas independientes: goles_local y goles_visitante.

```python
def separar_goles(df):
    """ Divide el resultado en goles locales y visitantes."""
    df[["goles_local", "goles_visitante"]] = (
        df["resultado"].str.split("-", expand=True).astype(int)
    )
```

´calculo_resultados(df)´

- Compara los goles para determinar victorias, empates y derrotas, calculando automáticamente los puntos.

1. compara si los goles del local son mayores que los del visitante = victoria
2. compara si los goles de local son iguales que los de visitante = empate
3. calcula si el equipo visitante gana = derrota
4. Aplica el sistema de puntuación clásico del fútbol: ganado se multiplica por 3 (* 3), empatado (1) suma 1 punto y Los partidos perdidos valen 0.

```python
def calculo_resultados(df):
    """ Calcula puntos, victorias, empates y derrotas automaticamente."""
    df["ganados_local"] = (df["goles_local"] > df["goles_visitante"]).astype(int)
    df["ganados_visitante"] = (df["goles_visitante"] > df["goles_local"]).astype(int)

    df["empatados_local"] = (df["goles_local"] == df["goles_visitante"]).astype(int)
    df["empatados_visitante"] = df["empatados_local"]

    df["perdidos_local"] = df["ganados_visitante"]
    df["perdidos_visitante"] = df["ganados_local"]

    df["puntos_local"] = df["ganados_local"] * 3 + df["empatados_local"]
    df["puntos_visitante"] = df["ganados_visitante"] * 3 + df["empatados_visitante"]
```

`estadisticas(df)`

1. Separa los datos en dos bloques (local y visitante),
2. los unifica en una sola tabla y los convierte a dataframe, los agrupa por nombre de equipo sumando todas sus métricas.
3. ordena la clasificación final de mayor a menor según los puntos.
4. retorna y resetea el index.

```python
def estadisticas(df):
    """ Unifica local/visitante, agrupa por equipo y genera la tabla de clasificación convertida a dataframe."""
    locales = df[["local", "goles_local", "goles_visitante", "ganados_local",
                  "empatados_local","perdidos_local","puntos_local"]]

    locales.columns=["equipo","goles_favor","goles_contra","ganados","empatados","perdidos","puntos"]

    visitante = df[["visitante","goles_visitante","goles_local","ganados_visitante",
                    "empatados_visitante","perdidos_visitante","puntos_visitante"]]

    visitante.columns=["equipo","goles_favor","goles_contra","ganados","empatados","perdidos","puntos"]

    equipos_unificados=pd.concat([locales,visitante], ignore_index=True)

    estadisticas_equipos_unificados=equipos_unificados.groupby("equipo").agg(
     goles_favor=("goles_favor", "sum"),
     goles_contra=("goles_contra", "sum"),
     partidos_jugados=("equipo", "count"),
     ganados=("ganados", "sum"),
     empatados=("empatados", "sum"),
     perdidos=("perdidos", "sum"),
     puntos=("puntos", "sum"),

    ).sort_values(by="puntos",ascending=False)

    return estadisticas_equipos_unificados.reset_index()
```

`main(path)`

- Función que controla todas las funciones

```python
def main(path):
    """ejecuta el flujo completo de análisis de la liga."""
    estadisticas_liga = cargar_partidos(path)
    separar_goles(estadisticas_liga)
    calculo_resultados(estadisticas_liga)
    estadisticas_equipos_unificados = estadisticas(estadisticas_liga)
    display(estadisticas_equipos_unificados)
```
